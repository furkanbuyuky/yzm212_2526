```python
from hmmlearn import hmm
import numpy as np
```

```python
# 'EV' kelimesi için model
model_ev = hmm.CategoricalHMM(n_components=2, init_params="")

model_ev.startprob_ = np.array([1.0, 0.0])

model_ev.transmat_ = np.array([
    [0.6, 0.4],
    [0.2, 0.8]
])

model_ev.emissionprob_ = np.array([
    [0.3, 0.7],   # e durumu: P(Low)=0.3, P(High)=0.7
    [0.9, 0.1]    # v durumu: P(Low)=0.9, P(High)=0.1
])
```

```python
# 'OKUL' kelimesi için model
model_okul = hmm.CategoricalHMM(n_components=4, init_params="")

model_okul.startprob_ = np.array([1.0, 0.0, 0.0, 0.0])

model_okul.transmat_ = np.array([
    [0.2, 0.8, 0.0, 0.0],
    [0.0, 0.2, 0.8, 0.0],
    [0.0, 0.0, 0.2, 0.8],
    [0.0, 0.0, 0.0, 1.0]
])

model_okul.emissionprob_ = np.array([
    [0.2, 0.8],   # o
    [0.3, 0.7],   # k
    [0.8, 0.2],   # u
    [0.7, 0.3]    # l
])
```

```python
def test_word(test_data, test_name):
    score_ev = model_ev.score(test_data)
    score_okul = model_okul.score(test_data)

    print(f"----- {test_name} -----")
    print("Test verisi:", test_data.T[0])
    print("EV Modeli Puanı   :", score_ev)
    print("OKUL Modeli Puanı :", score_okul)

    if score_ev > score_okul:
        print("Bu ses dizisi 'EV' kelimesine daha çok uymaktadır.\n")
    else:
        print("Bu ses dizisi 'OKUL' kelimesine daha çok uymaktadır.\n")
```

```python
# 3 farklı test verisi
test_data_1 = np.array([[1], [0]])              # [High, Low]
test_data_2 = np.array([[1], [1], [0], [0]])    # [High, High, Low, Low]
test_data_3 = np.array([[1], [0], [1], [0]])    # [High, Low, High, Low]

# testleri çalıştır
test_word(test_data_1, "Test 1")
test_word(test_data_2, "Test 2")
test_word(test_data_3, "Test 3")
```
